```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator

print("Qiskit libraries imported successfully.")
```

```python
q = QuantumRegister(16, 'q')
c = ClassicalRegister(16, 'c')

print("16 Qubits and 16 Classical Bits initialized successfully.")
```

```python
circuit = QuantumCircuit(q, c)

print("16-Qubit Quantum Circuit created successfully.")
```

```python
circuit.h(q)

print("Hadamard gate applied to all 16 qubits.")
```

```python
circuit.measure(q, c)

print("All 16 qubits measured successfully.")
```

```python
print(circuit.draw())
```

```python
simulator = AerSimulator()
job = simulator.run(circuit, shots=1)
result = job.result()
counts = result.get_counts()

print("Random 16-Qubit Number:")
random_binary = list(counts.keys())[0]
print(random_binary)
```

```python
random_decimal = int(random_binary, 2)

print("16-bit Random Binary Number:", random_binary)
print("Equivalent Decimal Number:", random_decimal)
```

```python
if len(random_binary) == 16:
    print("Successfully generated a 16-bit random number.")
else:
    print("Error: Output is not 16 bits.")
```
