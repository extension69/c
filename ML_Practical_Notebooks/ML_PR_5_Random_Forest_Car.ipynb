# ML Practical 5 – Random Forest Classification

**Aim:** Classify car acceptability using a Random Forest model and evaluate it with accuracy, precision, recall, F1-score and a confusion matrix.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    confusion_matrix, classification_report
)
```

```python
url = "https://raw.githubusercontent.com/retrowhiz/Car-Evaluation/master/car_evaluation.csv"
columns = ["buying", "maint", "doors", "persons", "lug_boot", "safety", "class"]
data = pd.read_csv(url, header=None, names=columns)

print("First 5 rows:")
print(data.head())
print("\nDataset Information:")
data.info()
print("\nDataset Description:")
print(data.describe(include="all"))
print("\nMissing Values:")
print(data.isnull().sum())
print("\nClass Distribution:")
print(data["class"].value_counts())
```

```python
label_encoders = {}
for column in data.columns:
    encoder = LabelEncoder()
    data[column] = encoder.fit_transform(data[column])
    label_encoders[column] = encoder

print("Encoded Dataset:")
print(data.head())

X = data.drop("class", axis=1)
y = data["class"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred, average="weighted"))
print("Recall:", recall_score(y_test, y_pred, average="weighted"))
print("F1 Score:", f1_score(y_test, y_pred, average="weighted"))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
importance = pd.Series(model.feature_importances_, index=X.columns)
print("Feature Importance:")
print(importance.sort_values(ascending=False))

importance.sort_values().plot(kind="barh", figsize=(8, 5))
plt.title("Random Forest Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Features")
plt.show()
```

```python
# ========== ML PRACTICAL 5: RANDOM FOREST CLASSIFICATION ==========

# AIM: Classify car acceptability using Random Forest.
# FLOW: Load -> Inspect -> Encode -> Split -> Train -> Predict -> Evaluate

# THEORY:
# Random Forest = supervised ensemble learning algorithm.
# Ensemble = combines multiple models for better predictions.
# Random Forest combines multiple Decision Trees.
# Decision Tree = makes decisions using if-else conditions.
# Classification = predicts categories/classes.
# Each tree predicts a class; majority voting gives final class.
# Bootstrap Sampling = random sampling WITH replacement.
# Feature Randomness = each split considers random feature subsets.
# WHY RANDOM FOREST? Reduces overfitting and improves stability.
# MORE TREES = usually stable accuracy, but more time/memory.

# LIBRARIES:
# pandas(pd) = dataset handling; matplotlib(plt) = graphs.
# train_test_split = divides training/testing data.
# LabelEncoder = converts categories into numbers.
# RandomForestClassifier = Random Forest classification model.
# sklearn.metrics = evaluates model performance.

# DATA LOADING:
# pd.read_csv(url,header=None,names=columns) = loads dataset.
# header=None = CSV has no header row.
# names=columns = assigns specified column names.
# head() = first 5 rows; info() = data types and counts.
# describe(include="all") = summary of all columns.
# isnull().sum() = counts missing values.
# value_counts() = frequency of each class.

# LABEL ENCODING:
# label_encoders={} = dictionary storing encoders.
# for column in data.columns = loops through every column.
# LabelEncoder() = creates categorical encoder.
# fit_transform(data[column]) = learns categories + converts to numbers.
# label_encoders[column]=encoder = saves encoder for each column.
# Example: low, med, high -> numerical labels.
# NOTE: LabelEncoder numbers are assigned by sorted class order.

# FEATURES & TARGET:
# X = data.drop("class",axis=1) = input features.
# y = data["class"] = target (car acceptability).
# axis=1 = columns; axis=0 = rows.

# TRAIN-TEST SPLIT:
# train_test_split(X,y,test_size=0.20,random_state=42,stratify=y)
# test_size=0.20 = 80% training, 20% testing.
# random_state=42 = reproducible random split.
# stratify=y = maintains class proportions in both sets.
# shape = (number of rows, number of columns).

# MODEL TRAINING:
# RandomForestClassifier(n_estimators=100,random_state=42)
# n_estimators=100 = creates 100 Decision Trees.
# random_state=42 = reproducible results.
# model.fit(X_train,y_train) = trains all trees.
# model.predict(X_test) = predicts car classes.
# FINAL PREDICTION = majority vote of the trees.

# EVALUATION:
# accuracy_score(actual,predicted) = overall correct predictions.
# precision_score() = how many predicted positives are correct.
# recall_score() = how many actual positives are detected.
# f1_score() = harmonic mean of precision and recall.
# average="weighted" = averages class scores by class sample count.
# confusion_matrix() = table of actual vs predicted classes.
# classification_report() = precision, recall, F1 and support.
# Support = number of actual samples in each class.

# FEATURE IMPORTANCE:
# model.feature_importances_ = importance of each input feature.
# pd.Series(values,index=X.columns) = labeled importance values.
# sort_values(ascending=False) = highest importance first.
# sort_values() = lowest importance first.
# plot(kind="barh",figsize=(8,5)) = horizontal bar graph.
# kind="barh" = horizontal bars; figsize = figure size.
# plt.title() = graph title; xlabel()/ylabel() = axis labels.
# plt.show() = displays graph.
# Higher importance = feature contributed more to tree splits.

# ========== 4 IMPORTANT VIVA QUESTIONS ==========

# Q1. Why use Random Forest over a single Decision Tree?
# Ans: It combines many trees, reducing overfitting and
#      improving prediction stability and accuracy.

# Q2. What happens when the number of trees increases?
# Ans: Predictions usually become more stable, but training
#      time and memory increase. Accuracy is not guaranteed to rise.

# Q3. What is Bootstrap Sampling?
# Ans: Randomly selecting training samples WITH replacement.
#      The same sample can be selected multiple times.

# Q4. Why introduce Feature Randomness?
# Ans: To make trees different, reduce correlation between trees,
#      and improve generalization.

# FINAL VIVA EXPLANATION:
# "We load the car evaluation dataset, inspect and encode
# categorical values, split data into 80% training and 20% testing,
# train a Random Forest with 100 Decision Trees, predict car
# acceptability, evaluate accuracy/precision/recall/F1,
# and visualize feature importance."
```
