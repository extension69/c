### 2. Import NLTK tools and download resources

Imports the tokenizer, English stop-word list, and Porter stemmer. The download loop fetches the language data required by the tokenizer and stop-word corpus so the later preprocessing steps can run.

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

# Download tokenizer resources and stop words
for resource in ("punkt", "punkt_tab", "stopwords"):
    nltk.download(resource, quiet=True)

print("NLTK resources are ready.")
```

### 3. Define sample text

Creates a small sample document containing several sentences. This text is the input on which the information-retrieval preprocessing steps will be demonstrated.

```python
text = """
This is the first text.
This is the second text. Hello! How are you?
This is the third text. We are learning Information Retrieval.
"""

print("Original Text:")
print(text)
```

### 4. Tokenize the text

Converts the text to lowercase and splits it into individual tokens (words and punctuation). Tokenization is a basic first step in preparing text for indexing or search.

```python
tokens = word_tokenize(text.lower())

print("Tokens:")
print(tokens)
```

### 5. Remove stop words

Loads common English stop words and keeps only alphabetic tokens that are not in that list. This reduces very frequent words such as “the” and “is” that often contribute little to search relevance.

```python
stop_words = set(stopwords.words("english"))

filtered_words = [
    word for word in tokens
    if word.isalpha() and word not in stop_words
]

print("After Stop Word Removal:")
print(filtered_words)
```

### 6. Apply stemming

Uses the Porter stemming algorithm to reduce words to approximate root forms (for example, “learning” may become “learn”). Stemming can group related word forms under a common token.

```python
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in filtered_words
]

print("After Stemming:")
print(stemmed_words)
```

### 7. Display preprocessing stages

Prints the original document, token list, stop-word-filtered words, and stemmed words together so the effect of each preprocessing step can be compared.

```python
print("Original Text:")
print(text)

print("\\nTokens:")
print(tokens)

print("\\nAfter Stop Word Removal:")
print(filtered_words)

print("\\nAfter Stemming:")
print(stemmed_words)
```
