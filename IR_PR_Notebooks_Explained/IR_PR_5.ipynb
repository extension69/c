### 2. Import libraries

Imports data and plotting tools, PCA for dimensionality reduction, AgglomerativeClustering for hierarchical clustering, StandardScaler for feature scaling, normalize for row-wise normalization, and linkage/dendrogram for hierarchical visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering
from sklearn.preprocessing import StandardScaler, normalize
from sklearn.metrics import silhouette_score
import scipy.cluster.hierarchy as shc
```

### 3. Load the customer dataset

Reads the credit-card customer dataset from the supplied CSV URL and displays the first rows. Internet access is needed to download the data.

```python
url = "https://raw.githubusercontent.com/j4-metrics/data_science_projects/main/cc_general.csv"

X = pd.read_csv(url)

print("Dataset loaded successfully!")
display(X.head())
```

### 4. Inspect dataset quality

Prints the dataset shape, column names, and missing-value counts to understand the available customer features and identify incomplete fields.

```python
print("Shape of dataset:", X.shape)

print("\\nColumn names:")
print(X.columns)

print("\\nMissing values:")
print(X.isnull().sum())
```

### 5. Remove customer identifier

Drops CUST_ID because it is an identifier rather than a meaningful numeric behaviour feature for distance-based clustering.

```python
# Remove the customer identifier because it is not a numeric behavioral feature.
X = X.drop("CUST_ID", axis=1)

print("CUST_ID column removed.")
display(X.head())
```

### 6. Handle missing values

Forward-fills missing values from the preceding row, then fills any remaining missing numeric values with their column medians. This ensures the numerical algorithms receive complete feature values.

```python
# Forward-fill missing values as in the practical.
X = X.ffill()

# Handle any remaining missing values (for example, if the first row was missing).
X = X.fillna(X.median(numeric_only=True))

print("Missing values handled.")
print("Remaining missing values:", int(X.isnull().sum().sum()))
```

### 7. Standardize features

Uses StandardScaler to center each feature around zero and scale it to unit variance. This prevents variables with large numeric ranges from dominating distance calculations.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Data standardized successfully.")
```

### 8. Normalize rows

Applies row-wise normalization to the standardized data so each customer vector has unit length. The result is stored as a DataFrame for subsequent analysis.

```python
X_normalized = normalize(X_scaled)
X_normalized = pd.DataFrame(X_normalized)

print("Data normalized successfully.")
display(X_normalized.head())
```

### 9. Apply PCA

Reduces the normalized feature set to two principal components. PCA creates new axes that capture as much variance as possible, allowing the high-dimensional customer data to be visualized in two dimensions.

```python
pca = PCA(n_components=2)
X_principal = pca.fit_transform(X_normalized)

X_principal = pd.DataFrame(X_principal, columns=["P1", "P2"])

print("PCA completed. Explained variance ratio:", pca.explained_variance_ratio_)
display(X_principal.head())
```

## Dendrogram

The original practical draws a hierarchical dendrogram. To keep plotting responsive, the visualization below uses a reproducible sample of up to 1,000 rows. The clustering experiments that follow use the full PCA dataset.

### 10. Create a dendrogram sample

Takes up to 1000 rows from the PCA output to keep the hierarchical dendrogram manageable. The sample is reproducible because random_state is fixed.

```python
sample_size = min(1000, len(X_principal))
dendrogram_sample = X_principal.sample(
    n=sample_size,
    random_state=42
)

plt.figure(figsize=(10, 5))
plt.title("Visualising the data")
dendrogram = shc.dendrogram(
    shc.linkage(dendrogram_sample, method="ward"),
    no_labels=True
)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.tight_layout()
plt.show()
```

### 11. Cluster customers with K=2

Fits agglomerative clustering with two clusters and plots the PCA points coloured by their assigned cluster.

```python
# Agglomerative Clustering for K = 2
ac2 = AgglomerativeClustering(n_clusters=2)

plt.figure(figsize=(4, 4))
plt.scatter(
    X_principal["P1"],
    X_principal["P2"],
    c=ac2.fit_predict(X_principal),
    cmap="rainbow",
    s=8
)
plt.xlabel("P1")
plt.ylabel("P2")
plt.title("Agglomerative Clustering (K = 2)")
plt.tight_layout()
plt.show()
```

### 12. Cluster customers with K=3

Repeats agglomerative clustering with three clusters and visualizes the resulting assignments in the two-component PCA space.

```python
# Agglomerative Clustering for K = 3
ac3 = AgglomerativeClustering(n_clusters=3)

plt.figure(figsize=(4, 4))
plt.scatter(
    X_principal["P1"],
    X_principal["P2"],
    c=ac3.fit_predict(X_principal),
    cmap="rainbow",
    s=8
)
plt.xlabel("P1")
plt.ylabel("P2")
plt.title("Agglomerative Clustering (K = 3)")
plt.tight_layout()
plt.show()
```

### 13. Cluster customers with K=4

Fits a four-cluster agglomerative model and displays the corresponding cluster labels on the PCA scatter plot.

```python
# Agglomerative Clustering for K = 4
ac4 = AgglomerativeClustering(n_clusters=4)

plt.figure(figsize=(4, 4))
plt.scatter(
    X_principal["P1"],
    X_principal["P2"],
    c=ac4.fit_predict(X_principal),
    cmap="rainbow",
    s=8
)
plt.xlabel("P1")
plt.ylabel("P2")
plt.title("Agglomerative Clustering (K = 4)")
plt.tight_layout()
plt.show()
```

### 14. Cluster customers with K=5

Fits a five-cluster agglomerative model and plots the five group assignments.

```python
# Agglomerative Clustering for K = 5
ac5 = AgglomerativeClustering(n_clusters=5)

plt.figure(figsize=(4, 4))
plt.scatter(
    X_principal["P1"],
    X_principal["P2"],
    c=ac5.fit_predict(X_principal),
    cmap="rainbow",
    s=8
)
plt.xlabel("P1")
plt.ylabel("P2")
plt.title("Agglomerative Clustering (K = 5)")
plt.tight_layout()
plt.show()
```

### 15. Cluster customers with K=6

Fits a six-cluster agglomerative model and plots the six group assignments.

```python
# Agglomerative Clustering for K = 6
ac6 = AgglomerativeClustering(n_clusters=6)

plt.figure(figsize=(4, 4))
plt.scatter(
    X_principal["P1"],
    X_principal["P2"],
    c=ac6.fit_predict(X_principal),
    cmap="rainbow",
    s=8
)
plt.xlabel("P1")
plt.ylabel("P2")
plt.title("Agglomerative Clustering (K = 6)")
plt.tight_layout()
plt.show()
```

### 16. Calculate silhouette scores

Evaluates candidate cluster counts from 2 through 6 using the silhouette score. A higher score generally indicates better separation and cohesion. The loop stores each score for comparison.

```python
k = [2, 3, 4, 5, 6]
cluster_models = {
    2: ac2,
    3: ac3,
    4: ac4,
    5: ac5,
    6: ac6
}

silhouette_scores = []

for clusters in k:
    labels = cluster_models[clusters].fit_predict(X_principal)
    score = silhouette_score(X_principal, labels)
    silhouette_scores.append(score)

print("Silhouette Scores:")
for clusters, score in zip(k, silhouette_scores):
    print(f"K = {clusters}: {score:.4f}")
```

### 17. Compare silhouette scores

Plots the silhouette score for each tested number of clusters so the candidate cluster counts can be compared visually.

```python
plt.figure(figsize=(6, 4))
plt.bar(k, silhouette_scores)
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score Comparison")
plt.xticks(k)
plt.tight_layout()
plt.show()
```

### 18. Select the best cluster count

Finds the cluster count with the highest silhouette score among the tested values and prints it as the selected number of clusters.

```python
optimal_k = k[int(np.argmax(silhouette_scores))]
print("Optimal number of clusters:", optimal_k)
```
