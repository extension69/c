### 2. Import libraries

QuantumCircuit defines the circuit; AerSimulator runs it; plot_histogram visualizes measurement frequencies; pyplot displays the plot.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

print("Libraries imported successfully.")
```

### 3. Create a 3-qubit circuit

Creates three qubits and three classical bits. The classical bits will hold the final measurement results.

```python
circuit = QuantumCircuit(3, 3)

print("Quantum circuit with 3 qubits and 3 classical bits created successfully.")
```

### 4. Apply an X gate

The X gate flips a computational-basis qubit: |0⟩ becomes |1⟩ and |1⟩ becomes |0⟩. The barrier is a visual separator in the circuit diagram.

```python
circuit.x(0)
circuit.barrier()

print("X gate applied to Qubit 0.")
```

### 5. Create superposition and entanglement

H puts qubit 1 into superposition. CX uses qubit 1 as control and qubit 2 as target, correlating the two qubits and creating entanglement.

```python
circuit.h(1)
circuit.cx(1, 2)

print("Entanglement created between Qubit 1 and Qubit 2.")
```

### 6. Apply more gates and measure two qubits

CX(0,1) applies a controlled-X from qubit 0 to qubit 1; H(0) changes qubit 0's basis. The selected measurements store qubits 0 and 1 in classical bits 0 and 1.

```python
circuit.cx(0, 1)
circuit.h(0)

circuit.barrier()
circuit.measure([0, 1], [0, 1])

print("Qubit 0 and Qubit 1 measured successfully.")
```

### 7. Apply controlled gates

The second CX acts from qubit 1 to qubit 2. CZ applies a controlled phase flip when both control and target are |1⟩. A barrier separates this stage.

```python
circuit.barrier()
circuit.cx(1, 2)
circuit.cz(0, 2)

print("Controlled-X and Controlled-Z gates applied.")
```

### 8. Measure the third qubit

Stores qubit 2's measurement in classical bit 2, completing the three-qubit measurement.

```python
circuit.measure(2, 2)

print("Qubit 2 measured successfully.")
```

### 9. Draw the circuit

Prints the full gate and measurement layout. Use it to verify the order and control-target connections.

```python
print(circuit.draw())
```

### 10. Simulate 1000 shots

Runs the circuit 1000 times. The resulting counts dictionary reports how often each three-bit measurement string occurred.

```python
simulator = AerSimulator()
job = simulator.run(circuit, shots=1000)
result = job.result()
counts = result.get_counts()

print("Simulation completed successfully.")
print("Measurement Results:")
print(counts)
```

### 11. Plot measurement counts

Creates a histogram with one bar per observed bitstring. Bar heights show the number of times each result appeared across the 1000 shots.

```python
fig = plot_histogram(counts, figsize=(5, 3))
plt.show()
```
