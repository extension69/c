### 2. Import libraries

Imports the quantum circuit and transpiler, AerSimulator, histogram plotting, and Matplotlib display support.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

print("Libraries imported successfully.")
```

### 3. Define initial and goal boards

Stores the initial 3×3 puzzle arrangement and the target arrangement. None represents the blank square. The cell prints each list as three rows for readability.

```python
initial_state = [2, 5, 3,
                 1, 8, 6,
                 4, 7, None]

goal_state = [1, 2, 3,
              4, 5, 6,
              7, 8, None]

print("Initial State:")
print(initial_state[:3])
print(initial_state[3:6])
print(initial_state[6:])

print("\\nGoal State:")
print(goal_state[:3])
print(goal_state[3:6])
print(goal_state[6:])
```

### 4. Create a nine-qubit circuit

Creates a circuit with nine qubits and nine classical bits, using one qubit position for each of the nine board positions.

```python
qc = QuantumCircuit(9, 9)

print("9-Qubit quantum circuit created successfully.")
```

### 5. Encode the initial board

Applies X to the qubit for each occupied tile and leaves the blank position in |0⟩. This is a simple occupancy encoding; it does not encode the tile numbers 1–8 distinctly.

```python
# Encode occupied tiles as |1> and the blank cell as |0>.
for i, tile in enumerate(initial_state):
    if tile is not None:
        qc.x(i)

qc.barrier()
print("Initial puzzle state encoded.")
```

### 6. Apply the listed quantum operations

Applies Hadamard gates to selected qubits and CX gates between neighboring positions, following the gate sequence shown in the supplied practical. These operations transform the encoded quantum state; they do not by themselves implement legal 8-puzzle moves or a complete search algorithm.

```python
# Apply the quantum operations shown in the supplied practical.
qc.h(0)
qc.h(1)

qc.cx(0, 1)
qc.cx(1, 2)

qc.h(3)
qc.h(4)
qc.h(5)

qc.cx(3, 4)
qc.cx(4, 5)

qc.h(6)
qc.h(7)

qc.cx(6, 7)
qc.cx(7, 8)

qc.barrier()
print("Quantum operations applied successfully.")
```

### 7. Measure and draw the circuit

Measures all nine qubits into nine classical bits and prints the circuit diagram, showing the encoding, gates, and measurements.

```python
qc.measure(range(9), range(9))

print("All qubits measured successfully.")
print(qc.draw())
```

### 8. Simulate 1024 shots

Transpiles the circuit for AerSimulator and executes it 1024 times. The counts dictionary records how often each nine-bit measurement string appears.

```python
simulator = AerSimulator()
compiled_qc = transpile(qc, simulator)

job = simulator.run(compiled_qc, shots=1024)
result = job.result()
counts = result.get_counts()

print("Simulation completed successfully.")
print("\\nMeasurement Results:")
print(counts)
```

### 9. Find the most frequent bitstring

Selects the key with the largest count and prints it along with its occurrence count. This is the most frequent measurement result, not necessarily a verified solution to the 8-puzzle.

```python
solution = max(counts, key=counts.get)

print("Most probable quantum state:")
print(solution)

print("\\nNumber of occurrences:")
print(counts[solution])
```

### 10. Plot measurement distribution

Displays the measurement counts as a histogram. With nine measured bits there may be many possible outcomes, so labels can become crowded.

```python
fig = plot_histogram(counts, figsize=(10, 4), sort='value')
plt.show()
```
