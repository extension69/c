### 2. Import libraries

QuantumRegister stores qubits, ClassicalRegister stores measurement bits, and QuantumCircuit combines them into a circuit. AerSimulator executes the circuit on a local simulator.

```python
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister
from qiskit_aer import AerSimulator

print("Qiskit libraries imported successfully.")
```

### 3. Create registers

Creates 16 quantum bits and 16 classical bits. The classical bits store the measurement results from the corresponding qubits.

```python
q = QuantumRegister(16, 'q')
c = ClassicalRegister(16, 'c')

print("16 Qubits and 16 Classical Bits initialized successfully.")
```

### 4. Create the circuit

Combines the quantum and classical registers into one 16-qubit circuit. Gates will be added to this circuit in the following cells.

```python
circuit = QuantumCircuit(q, c)

print("16-Qubit Quantum Circuit created successfully.")
```

### 5. Apply Hadamard gates

The Hadamard (H) gate puts each qubit into an equal superposition of |0⟩ and |1⟩. Measuring these qubits produces a random-looking 16-bit string.

```python
circuit.h(q)

print("Hadamard gate applied to all 16 qubits.")
```

### 6. Measure the qubits

Measures every qubit and writes its result into the matching classical bit. Measurement converts the quantum state into classical data.

```python
circuit.measure(q, c)

print("All 16 qubits measured successfully.")
```

### 7. Draw the circuit

Displays the circuit diagram so you can inspect the H gates and measurement operations applied to all 16 qubits.

```python
print(circuit.draw())
```

### 8. Run the simulation

AerSimulator runs the circuit once because shots=1. get_counts() returns a dictionary whose key is the measured bitstring and whose value is how often it occurred.

```python
simulator = AerSimulator()
job = simulator.run(circuit, shots=1)
result = job.result()
counts = result.get_counts()

print("Random 16-Qubit Number:")
random_binary = list(counts.keys())[0]
print(random_binary)
```

### 9. Convert binary to decimal

int(random_binary, 2) interprets the measured bitstring as a base-2 integer and converts it to its decimal equivalent.

```python
random_decimal = int(random_binary, 2)

print("16-bit Random Binary Number:", random_binary)
print("Equivalent Decimal Number:", random_decimal)
```

### 10. Check output length

Checks that the simulator returned exactly 16 bits. This is a simple validation that the output has the requested width.

```python
if len(random_binary) == 16:
    print("Successfully generated a 16-bit random number.")
else:
    print("Error: Output is not 16 bits.")
```
