# IR Practical 2: Inverted Index and Boolean AND Search
**Aim:** Build an inverted index for a collection of documents and retrieve documents containing all query terms.

```python
import re
from collections import defaultdict
```

```python
documents = [
    "This is the first text.",
    "This is the second text. Hello! How are you?",
    "This is the third text. We are learning Information Retrieval."
]

print("Documents:")
for i, document in enumerate(documents, start=1):
    print(f"Document {i}: {document}")
```

```python
inverted_index = defaultdict(list)

for doc_id, document in enumerate(documents, start=1):
    document = document.lower()
    words = re.findall(r"\\b[a-z]+\\b", document)

    # Add each document ID once for every unique word in that document
    for word in set(words):
        inverted_index[word].append(doc_id)

print("Inverted Index:")
for word in sorted(inverted_index):
    print(word, ":", sorted(inverted_index[word]))
```

```python
print("----- INVERTED INDEX -----")
for word in sorted(inverted_index):
    print(f"{word:15} -> {sorted(inverted_index[word])}")
```

```python
def search(query):
    """Return document IDs containing every term in the query (AND search)."""
    query_words = re.findall(r"\\b[a-z]+\\b", query.lower())

    if not query_words:
        return []

    result_sets = [
        set(inverted_index.get(word, []))
        for word in query_words
    ]

    result = set.intersection(*result_sets)
    return sorted(result)
```

```python
query = input("Enter search query: ")
result = search(query)

print("\\nQuery:", query)
if result:
    print("Matching Documents:", result)
else:
    print("No matching documents found.")
```

```python
# Example searches shown in the practical
for query in ("text", "hello", "information retrieval", "missingterm"):
    result = search(query)
    print(f"Query: {query!r}")
    print("Matching Documents:", result if result else "No matching documents found.")
    print()
```

```python
print("Result:")
print("The inverted index was successfully created.")
print("Documents were successfully retrieved using the search query.")
```
