# IR Practical 5: Customer Segmentation using PCA and Agglomerative Clustering
**Aim:** Standardize and normalize customer data, reduce dimensions using PCA, visualize a dendrogram, perform Agglomerative Clustering for K = 2 to 6, and compare clusters using the Silhouette Score.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.cluster import AgglomerativeClustering
from sklearn.preprocessing import StandardScaler, normalize
from sklearn.metrics import silhouette_score
import scipy.cluster.hierarchy as shc
```

```python
url = "https://raw.githubusercontent.com/j4-metrics/data_science_projects/main/cc_general.csv"

X = pd.read_csv(url)

print("Dataset loaded successfully!")
display(X.head())
```

```python
print("Shape of dataset:", X.shape)

print("\\nColumn names:")
print(X.columns)

print("\\nMissing values:")
print(X.isnull().sum())
```

```python
# Remove the customer identifier because it is not a numeric behavioral feature.
X = X.drop("CUST_ID", axis=1)

print("CUST_ID column removed.")
display(X.head())
```

```python
# Forward-fill missing values as in the practical.
X = X.ffill()

# Handle any remaining missing values (for example, if the first row was missing).
X = X.fillna(X.median(numeric_only=True))

print("Missing values handled.")
print("Remaining missing values:", int(X.isnull().sum().sum()))
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Data standardized successfully.")
```

```python
X_normalized = normalize(X_scaled)
X_normalized = pd.DataFrame(X_normalized)

print("Data normalized successfully.")
display(X_normalized.head())
```

```python
pca = PCA(n_components=2)
X_principal = pca.fit_transform(X_normalized)

X_principal = pd.DataFrame(X_principal, columns=["P1", "P2"])

print("PCA completed. Explained variance ratio:", pca.explained_variance_ratio_)
display(X_principal.head())
```

## Dendrogram

The original practical draws a hierarchical dendrogram. To keep plotting responsive, the visualization below uses a reproducible sample of up to 1,000 rows. The clustering experiments that follow use the full PCA dataset.

```python
sample_size = min(1000, len(X_principal))
dendrogram_sample = X_principal.sample(
    n=sample_size,
    random_state=42
)

plt.figure(figsize=(10, 5))
plt.title("Visualising the data")
dendrogram = shc.dendrogram(
    shc.linkage(dendrogram_sample, method="ward"),
    no_labels=True
)
plt.xlabel("Data Points")
plt.ylabel("Distance")
plt.tight_layout()
plt.show()
```

```python
# Agglomerative Clustering for K = 2
ac2 = AgglomerativeClustering(n_clusters=2)

plt.figure(figsize=(4, 4))
plt.scatter(
    X_principal["P1"],
    X_principal["P2"],
    c=ac2.fit_predict(X_principal),
    cmap="rainbow",
    s=8
)
plt.xlabel("P1")
plt.ylabel("P2")
plt.title("Agglomerative Clustering (K = 2)")
plt.tight_layout()
plt.show()
```

```python
# Agglomerative Clustering for K = 3
ac3 = AgglomerativeClustering(n_clusters=3)

plt.figure(figsize=(4, 4))
plt.scatter(
    X_principal["P1"],
    X_principal["P2"],
    c=ac3.fit_predict(X_principal),
    cmap="rainbow",
    s=8
)
plt.xlabel("P1")
plt.ylabel("P2")
plt.title("Agglomerative Clustering (K = 3)")
plt.tight_layout()
plt.show()
```

```python
# Agglomerative Clustering for K = 4
ac4 = AgglomerativeClustering(n_clusters=4)

plt.figure(figsize=(4, 4))
plt.scatter(
    X_principal["P1"],
    X_principal["P2"],
    c=ac4.fit_predict(X_principal),
    cmap="rainbow",
    s=8
)
plt.xlabel("P1")
plt.ylabel("P2")
plt.title("Agglomerative Clustering (K = 4)")
plt.tight_layout()
plt.show()
```

```python
# Agglomerative Clustering for K = 5
ac5 = AgglomerativeClustering(n_clusters=5)

plt.figure(figsize=(4, 4))
plt.scatter(
    X_principal["P1"],
    X_principal["P2"],
    c=ac5.fit_predict(X_principal),
    cmap="rainbow",
    s=8
)
plt.xlabel("P1")
plt.ylabel("P2")
plt.title("Agglomerative Clustering (K = 5)")
plt.tight_layout()
plt.show()
```

```python
# Agglomerative Clustering for K = 6
ac6 = AgglomerativeClustering(n_clusters=6)

plt.figure(figsize=(4, 4))
plt.scatter(
    X_principal["P1"],
    X_principal["P2"],
    c=ac6.fit_predict(X_principal),
    cmap="rainbow",
    s=8
)
plt.xlabel("P1")
plt.ylabel("P2")
plt.title("Agglomerative Clustering (K = 6)")
plt.tight_layout()
plt.show()
```

```python
k = [2, 3, 4, 5, 6]
cluster_models = {
    2: ac2,
    3: ac3,
    4: ac4,
    5: ac5,
    6: ac6
}

silhouette_scores = []

for clusters in k:
    labels = cluster_models[clusters].fit_predict(X_principal)
    score = silhouette_score(X_principal, labels)
    silhouette_scores.append(score)

print("Silhouette Scores:")
for clusters, score in zip(k, silhouette_scores):
    print(f"K = {clusters}: {score:.4f}")
```

```python
plt.figure(figsize=(6, 4))
plt.bar(k, silhouette_scores)
plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score Comparison")
plt.xticks(k)
plt.tight_layout()
plt.show()
```

```python
optimal_k = k[int(np.argmax(silhouette_scores))]
print("Optimal number of clusters:", optimal_k)
```
